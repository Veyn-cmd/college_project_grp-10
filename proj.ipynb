```python
numeric_df = df.select_dtypes(include="number")

corr = numeric_df.corr()

plt.figure(figsize=(15, 10))
sns.heatmap(corr, cmap="coolwarm", center=0)
plt.title("Correlation Heatmap")
plt.show()
```

# Correlation analysis


```python
import numpy as np

corr_pairs = (
    corr.where(~np.eye(corr.shape[0], dtype=bool))
        .stack()
        .sort_values(key=abs, ascending=False)
)

print(corr_pairs.head(20))
```

The strongest correlation is between Curricular units 1st sem (credited) and Curricular units 2nd sem (credited), with a correlation of 0.945. This indicates a very strong positive relationship between credited units across the two semesters.

Curricular units 1st sem (enrolled) and Curricular units 2nd sem (enrolled) also have a very strong positive correlation of 0.943.

Mother's occupation and Father's occupation have a strong positive correlation of 0.910.

There is also a strong relationship between semester academic-performance variables. For example, Curricular units 1st sem (approved) and Curricular units 2nd sem (approved) have a correlation of 0.904, while the first- and second-semester grades have a correlation of 0.837.

Nationality and International have a correlation of 0.791, while first- and second-semester evaluations have a correlation of 0.779.

Overall, the correlation matrix shows that students' academic activity and performance tend to be strongly related across the two semesters.
